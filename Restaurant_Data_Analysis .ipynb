# Task 1: Top 3 Cuisines percentage(%)

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv('C:/Users/golut/Downloads/Dataset .csv')
```

### Data Cleaning
After checking the dataset, I found 9 missing values in the "Cuisines" column. 
I have removed these values in the next step to ensure accurate analysis.

```python
df = df.dropna(subset=["Cuisines"])
cuisines = df["Cuisines"].str.split(", ")
all_cuisines = cuisines.explode()
top_cuisines = all_cuisines.value_counts().head(3)
percentage = (top_cuisines / len(df)) * 100
percentage=percentage.round(2)
print(percentage)
plt.bar(percentage.index, percentage.values, color=['skyblue'])
plt.title("Top 3 Cuisines Distribution (%)",fontsize=14)
plt.ylabel("Percentage",fontsize=16)
plt.xlabel("City",fontsize=18)
plt.show()
```

## Conclusion

* North Indian is the most popular cuisine with a 41.50% share in the dataset.

* Top 3 Cuisines: North Indian, Chinese (28.66%), and Fast Food (20.81%) dominate the market.

* These three cuisines capture over 90% of the customer preference, showing a very strong demand for traditional and quick-service meals.



# Task 2: City-wise Restaurant and Rating Analysis

### * City-wise Restaurant Count

```python
top_cities = df['City'].value_counts().head(5)
print(top_cities)
plt.figure(figsize=(5,3))
top_cities.plot(kind='bar')
plt.title("Top 5 Cities by Number of Restaurants", fontsize=14)
plt.xlabel("City", fontsize=14)
plt.ylabel("Number of Restaurants", fontsize=12)
plt.xticks(rotation=45)
plt.show()
```

### * Average Rating by City

```python
avg_rating = df.groupby('City')['Aggregate rating'].mean().sort_values(ascending=False).head(5).round(2).reset_index()
avg_rating.index += 1
avg_rating
```

## Conclusion
* New Delhi is the biggest market with 5,473 restaurants, showing huge demand and high competition.

* Inner City has the best quality with a 4.9 average rating.
Insight: Where restaurants are fewer, ratings are higher (better quality control).

# Task 3: Price Range Distribution Analysis

## *Price Range Distribution Visualization

```python
price_counts = df['Price range'].value_counts()
print(price_counts)
price_counts.plot(kind='bar')
plt.title('Price Range Distribution',fontsize=16)
plt.xlabel('Price Range',fontsize=16)
plt.ylabel('Number of Restaurants',fontsize=14)
plt.xticks(rotation=0)
plt.show()
```

## *Percentage Distribution of Price Range

```python
price_percentage = (price_counts / len(df)) * 100
price_percentage.round(2).astype(str) + '%'
```

## Conclusion
* The analysis shows that most restaurants are in the lower price ranges. This means customers prefer affordable food. High-price restaurants are fewer, showing less demand for expensive dining.

# Task 4: Online Delivery

## *Percentage of restaurants offering online delivery

```python
delivery_counts = df['Has Online delivery'].value_counts()
delivery_counts.plot(kind='pie', autopct='%1.1f%%')
plt.title('Online Delivery Distribution')
plt.ylabel('') 
plt.show()
```

## * Compare average ratings (With vs Without Online Delivery)

```python
delivery_rating = df.groupby('Has Online delivery')['Aggregate rating'].mean().round(2)
print(delivery_rating)
colors_name = ['slategray', 'turquoise']
delivery_rating.plot(kind='bar',color=colors_name)
plt.title('Average Rating: With vs Without Online Delivery',fontsize=14)
plt.xlabel('Online Delivery',fontsize=16)
plt.ylabel('Average Rating',fontsize=16)
plt.xticks(rotation=0)
plt.show()
```

## Conclusion

The analysis shows that restaurants offering online delivery have a significantly higher average rating (3.25) compared to those without online delivery (2.46). This suggests that online delivery services may positively influence customer satisfaction and overall restaurant ratings. Therefore, providing online delivery can be an effective strategy for improving customer experience and enhancing restaurant performance.
